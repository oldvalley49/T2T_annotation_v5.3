```python
from mjol.base import *
from mjol.gan import *
from mjol.tools import *
from collections import defaultdict
```

```python
prev_fn = "/ccb/salz4-3/hji20/chm13/chm13v2.0_RefSeq_Liftoff_v5.2.gff3"
prev_chm13 = GAn(
    file_name = prev_fn,
    file_fmt = 'gff'
)
prev_chm13.build_db()
```

```python
curr_fn = "/ccb/salz4-3/hji20/chm13/out/final.sorted.gff"
curr_chm13 = GAn(
    file_name = curr_fn,
    file_fmt = 'gff'
)
curr_chm13.build_db()
```

```python
cn_by_gtype_prev = defaultdict(int)
for f in prev_chm13.features.values():
    if f.puid:
        continue
    pc = False
    for c in f.children:
        for gc in c.children:
            if gc.feature_type == "CDS":
                pc = True
                break
        if pc:
            break
    if pc:
        cn_by_gtype_prev['protein_coding'] += 1
    else:
        if 'gene_biotype' in f.attributes:
            gtype = f.attributes['gene_biotype']
            cn_by_gtype_prev[gtype] += 1
        else:
            cn_by_gtype_prev['unknown'] += 1
cn_by_gtype_prev
```

```python
ctr = 0
for gtype in cn_by_gtype_prev:
    if 'pseudogene' in gtype:
        ctr += cn_by_gtype_prev[gtype]
ctr
```

```python
sum(cn_by_gtype_prev.values()) - ctr - cn_by_gtype_prev['lncRNA'] - cn_by_gtype_prev['protein_coding']
```

```python
sum(cn_by_gtype_prev.values())
```
